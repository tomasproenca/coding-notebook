```python
import random

def generate_numbers(size=10,min_value=0,max_value=100,unique=False,seed=None):

    """
    Generate a list of random integers.

    Parameters
    ----------
    size : int
        Number of elements to generate.
    min_value : int
        Minimum possible value.
    max_value : int
        Maximum possible value.
    unique : bool
        Whether all generated numbers must be unique.
    seed : int | None
        Optional seed for reproducible results.

    Returns
    -------
    list[int]
        Generated list of integers.
    """

    if size < 0:
        raise ValueError("size must be >= 0")

    if min_value > max_value:
        raise ValueError("min_value must be <= max_value")

    if unique and size > (max_value - min_value + 1):
        raise ValueError(
            "Cannot generate enough unique numbers within the given range"
        )

    if seed is not None:
        random.seed(seed)

    if unique:
        return random.sample(
            range(min_value, max_value + 1),
            size
        )

    return [
        random.randint(min_value, max_value)
        for _ in range(size)
    ]
```

## Basic

```python
# Bubble Sort

def bubble_sort(numbers):
    n = len(numbers)

    while n > 1:
        i = 0

        while i < n - 1:
            if numbers[i] > numbers[i + 1]:
                numbers[i], numbers[i + 1] = numbers[i + 1], numbers[i]

            i += 1

        n -= 1

    return numbers
```

```python
# Selection Sort

def selection_sort(numbers):

    for i in range(len(numbers)):
        min_index = i

        for j in range(i + 1, len(numbers)):
            if numbers[j] < numbers[min_index]:
                min_index = j

        numbers[i], numbers[min_index] = numbers[min_index], numbers[i]

    return numbers
```

```python
# Insertion Sort

def insertion_sort(numbers):

    for i in range(1, len(numbers)):
        current = numbers[i]
        j = i - 1

        while j >= 0 and numbers[j] > current:
            numbers[j + 1] = numbers[j]
            j -= 1

        numbers[j + 1] = current

    return numbers
```

```python
# Cocktail Shaker Sort

def cocktail_shaker_sort(numbers):
    start = 0
    end = len(numbers) - 1
    swapped = True

    while swapped:
        swapped = False

        for i in range(start, end):
            if numbers[i] > numbers[i + 1]:
                numbers[i], numbers[i + 1] = numbers[i + 1], numbers[i]
                swapped = True

        if not swapped:
            break

        end -= 1
        swapped = False

        for i in range(end - 1, start - 1, -1):
            if numbers[i] > numbers[i + 1]:
                numbers[i], numbers[i + 1] = numbers[i + 1], numbers[i]
                swapped = True

        start += 1

    return numbers
```

```python
# Gnome Sort

def gnome_sort(numbers):
    i = 0

    while i < len(numbers):
        if i == 0 or numbers[i] >= numbers[i - 1]:
            i += 1
        else:
            numbers[i], numbers[i - 1] = numbers[i - 1], numbers[i]
            i -= 1

    return numbers
```

```python
# Comb Sort

def comb_sort(numbers):
    gap = len(numbers)
    shrink = 1.3
    swapped = True

    while gap > 1 or swapped:
        gap = int(gap / shrink)

        if gap < 1:
            gap = 1

        swapped = False

        for i in range(len(numbers) - gap):
            if numbers[i] > numbers[i + gap]:
                numbers[i], numbers[i + gap] = numbers[i + gap], numbers[i]
                swapped = True

    return numbers
```

```python
# Shell Sort

def shell_sort(numbers):
    gap = len(numbers) // 2

    while gap > 0:

        for i in range(gap, len(numbers)):
            current = numbers[i]
            j = i

            while j >= gap and numbers[j - gap] > current:
                numbers[j] = numbers[j - gap]
                j -= gap

            numbers[j] = current

        gap //= 2

    return numbers
```

## Divide and Conquer

```python
# Merge Sort

def merge_sort(numbers):

    if len(numbers) <= 1:
        return numbers

    middle = len(numbers) // 2

    left = merge_sort(numbers[:middle])
    right = merge_sort(numbers[middle:])

    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result
```

```python
# Quick Sort

def quick_sort(numbers):

    if len(numbers) <= 1:
        return numbers

    pivot = numbers[len(numbers) // 2]

    left = []
    middle = []
    right = []

    for number in numbers:
        if number < pivot:
            left.append(number)
        elif number > pivot:
            right.append(number)
        else:
            middle.append(number)

    return quick_sort(left) + middle + quick_sort(right)
```

```python
# Heap Sort

def heap_sort(numbers):

    def heapify(numbers, heap_size, root):

        largest = root
        left = 2 * root + 1
        right = 2 * root + 2

        if left < heap_size and numbers[left] > numbers[largest]:
            largest = left

        if right < heap_size and numbers[right] > numbers[largest]:
            largest = right

        if largest != root:
            numbers[root], numbers[largest] = numbers[largest], numbers[root]
            heapify(numbers, heap_size, largest)

    n = len(numbers)

    for i in range(n // 2 - 1, -1, -1):
        heapify(numbers, n, i)

    for i in range(n - 1, 0, -1):
        numbers[0], numbers[i] = numbers[i], numbers[0]
        heapify(numbers, i, 0)

    return numbers
```

```python
# Tim Sort

def tim_sort(numbers):

    RUN = 32
    n = len(numbers)

    def insertion_sort(numbers, left, right):

        for i in range(left + 1, right + 1):
            current = numbers[i]
            j = i - 1

            while j >= left and numbers[j] > current:
                numbers[j + 1] = numbers[j]
                j -= 1

            numbers[j + 1] = current

    def merge(numbers, left, middle, right):

        left_part = numbers[left:middle + 1]
        right_part = numbers[middle + 1:right + 1]

        i = 0
        j = 0
        k = left

        while i < len(left_part) and j < len(right_part):
            if left_part[i] <= right_part[j]:
                numbers[k] = left_part[i]
                i += 1
            else:
                numbers[k] = right_part[j]
                j += 1

            k += 1

        while i < len(left_part):
            numbers[k] = left_part[i]
            i += 1
            k += 1

        while j < len(right_part):
            numbers[k] = right_part[j]
            j += 1
            k += 1

    for left in range(0, n, RUN):
        right = min(left + RUN - 1, n - 1)
        insertion_sort(numbers, left, right)

    size = RUN

    while size < n:

        for left in range(0, n, 2 * size):

            middle = min(left + size - 1, n - 1)
            right = min(left + 2 * size - 1, n - 1)

            if middle < right:
                merge(numbers, left, middle, right)

        size *= 2

    return numbers
```

```python
# IntroSort

def intro_sort(numbers):

    def insertion_sort(numbers, low, high):

        for i in range(low + 1, high + 1):
            current = numbers[i]
            j = i - 1

            while j >= low and numbers[j] > current:
                numbers[j + 1] = numbers[j]
                j -= 1

            numbers[j + 1] = current

    def heapify(numbers, heap_size, root, offset):

        largest = root
        left = 2 * root + 1
        right = 2 * root + 2

        if (
            left < heap_size
            and numbers[offset + left] > numbers[offset + largest]
        ):
            largest = left

        if (
            right < heap_size
            and numbers[offset + right] > numbers[offset + largest]
        ):
            largest = right

        if largest != root:
            numbers[offset + root], numbers[offset + largest] = (
                numbers[offset + largest],
                numbers[offset + root]
            )

            heapify(numbers, heap_size, largest, offset)

    def heap_sort_range(numbers, low, high):

        heap_size = high - low + 1

        for i in range(heap_size // 2 - 1, -1, -1):
            heapify(numbers, heap_size, i, low)

        for i in range(heap_size - 1, 0, -1):
            numbers[low], numbers[low + i] = (
                numbers[low + i],
                numbers[low]
            )

            heapify(numbers, i, 0, low)

    def partition(numbers, low, high):

        pivot = numbers[high]
        i = low

        for j in range(low, high):

            if numbers[j] <= pivot:
                numbers[i], numbers[j] = numbers[j], numbers[i]
                i += 1

        numbers[i], numbers[high] = numbers[high], numbers[i]

        return i

    def sort(numbers, low, high, depth_limit):

        if high - low <= 16:
            insertion_sort(numbers, low, high)
            return

        if depth_limit == 0:
            heap_sort_range(numbers, low, high)
            return

        pivot = partition(numbers, low, high)

        sort(numbers, low, pivot - 1, depth_limit - 1)
        sort(numbers, pivot + 1, high, depth_limit - 1)

    n = len(numbers)

    if n <= 1:
        return numbers

    depth_limit = 2 * (n.bit_length() - 1)

    sort(numbers, 0, n - 1, depth_limit)

    return numbers
```

## Counting

```python
# Counting Sort

def counting_sort(numbers):

    if not numbers:
        return numbers

    min_value = min(numbers)
    max_value = max(numbers)

    count = [0] * (max_value - min_value + 1)

    for number in numbers:
        count[number - min_value] += 1

    index = 0

    for i in range(len(count)):
        while count[i] > 0:
            numbers[index] = i + min_value
            index += 1
            count[i] -= 1

    return numbers
```

```python
# Bucket Sort

def bucket_sort(numbers):

    if len(numbers) <= 1:
        return numbers

    min_value = min(numbers)
    max_value = max(numbers)

    bucket_count = len(numbers)
    buckets = [[] for _ in range(bucket_count)]

    value_range = max_value - min_value

    for number in numbers:

        if value_range == 0:
            index = 0
        else:
            index = int(
                (number - min_value)
                * (bucket_count - 1)
                / value_range
            )

        buckets[index].append(number)

    index = 0

    for bucket in buckets:

        for i in range(1, len(bucket)):
            current = bucket[i]
            j = i - 1

            while j >= 0 and bucket[j] > current:
                bucket[j + 1] = bucket[j]
                j -= 1

            bucket[j + 1] = current

        for number in bucket:
            numbers[index] = number
            index += 1

    return numbers
```

```python
# Radix Sort

def radix_sort(numbers):

    if len(numbers) <= 1:
        return numbers

    def sort_positive(numbers):

        if not numbers:
            return numbers

        max_value = max(numbers)
        exponent = 1

        while max_value // exponent > 0:

            buckets = [[] for _ in range(10)]

            for number in numbers:
                digit = (number // exponent) % 10
                buckets[digit].append(number)

            index = 0

            for bucket in buckets:
                for number in bucket:
                    numbers[index] = number
                    index += 1

            exponent *= 10

        return numbers

    positives = []
    negatives = []

    for number in numbers:
        if number >= 0:
            positives.append(number)
        else:
            negatives.append(-number)

    sort_positive(positives)
    sort_positive(negatives)

    index = 0

    for number in reversed(negatives):
        numbers[index] = -number
        index += 1

    for number in positives:
        numbers[index] = number
        index += 1

    return numbers
```

```python
# Pigeonhole Sort

def pigeonhole_sort(numbers):

    if not numbers:
        return numbers

    min_value = min(numbers)
    max_value = max(numbers)

    holes = [0] * (max_value - min_value + 1)

    for number in numbers:
        holes[number - min_value] += 1

    index = 0

    for i in range(len(holes)):

        while holes[i] > 0:
            numbers[index] = i + min_value
            index += 1
            holes[i] -= 1

    return numbers
```

```python
# Tree Sort

def tree_sort(numbers):

    class Node:

        def __init__(self, value):
            self.value = value
            self.left = None
            self.right = None

    def insert(root, value):

        if root is None:
            return Node(value)

        if value < root.value:
            root.left = insert(root.left, value)
        else:
            root.right = insert(root.right, value)

        return root

    def inorder(root, result):

        if root is None:
            return

        inorder(root.left, result)
        result.append(root.value)
        inorder(root.right, result)

    if not numbers:
        return numbers

    root = None

    for number in numbers:
        root = insert(root, number)

    result = []
    inorder(root, result)

    numbers[:] = result

    return numbers
```

```python
# Cycle Sort

def cycle_sort(numbers):

    n = len(numbers)

    for cycle_start in range(n - 1):

        item = numbers[cycle_start]
        position = cycle_start

        for i in range(cycle_start + 1, n):
            if numbers[i] < item:
                position += 1

        if position == cycle_start:
            continue

        while item == numbers[position]:
            position += 1

        numbers[position], item = item, numbers[position]

        while position != cycle_start:

            position = cycle_start

            for i in range(cycle_start + 1, n):
                if numbers[i] < item:
                    position += 1

            while item == numbers[position]:
                position += 1

            numbers[position], item = item, numbers[position]

    return numbers
```

```python
# Bitonic Sort

def bitonic_sort(numbers):

    if len(numbers) <= 1:
        return numbers

    original_size = len(numbers)

    size = 1

    while size < original_size:
        size *= 2

    padding = [float("inf")] * (size - original_size)
    values = numbers + padding

    def compare_and_swap(values, i, j, ascending):

        if ascending == (values[i] > values[j]):
            values[i], values[j] = values[j], values[i]

    def bitonic_merge(values, low, count, ascending):

        if count <= 1:
            return

        distance = count // 2

        for i in range(low, low + distance):
            compare_and_swap(
                values,
                i,
                i + distance,
                ascending
            )

        bitonic_merge(values, low, distance, ascending)
        bitonic_merge(values, low + distance, distance, ascending)

    def sort(values, low, count, ascending):

        if count <= 1:
            return

        half = count // 2

        sort(values, low, half, True)
        sort(values, low + half, half, False)

        bitonic_merge(values, low, count, ascending)

    sort(values, 0, size, True)

    numbers[:] = values[:original_size]

    return numbers
```

```python
# Odd-Even Sort

def odd_even_sort(numbers):

    n = len(numbers)
    sorted_flag = False

    while not sorted_flag:

        sorted_flag = True

        for i in range(1, n - 1, 2):

            if numbers[i] > numbers[i + 1]:
                numbers[i], numbers[i + 1] = (
                    numbers[i + 1],
                    numbers[i]
                )

                sorted_flag = False

        for i in range(0, n - 1, 2):

            if numbers[i] > numbers[i + 1]:
                numbers[i], numbers[i + 1] = (
                    numbers[i + 1],
                    numbers[i]
                )

                sorted_flag = False

    return numbers
```

```python
# Stooge Sort

def stooge_sort(numbers):

    def sort(numbers, low, high):

        if numbers[low] > numbers[high]:
            numbers[low], numbers[high] = (
                numbers[high],
                numbers[low]
            )

        if high - low + 1 > 2:

            third = (high - low + 1) // 3

            sort(numbers, low, high - third)
            sort(numbers, low + third, high)
            sort(numbers, low, high - third)

    if numbers:
        sort(numbers, 0, len(numbers) - 1)

    return numbers
```

```python
# Bogo Sort

def bogo_sort(numbers):

    def is_sorted(numbers):

        for i in range(len(numbers) - 1):
            if numbers[i] > numbers[i + 1]:
                return False

        return True

    while not is_sorted(numbers):
        random.shuffle(numbers)

    return numbers
```

## I Don't Know

```python
# Pancake Sort

def pancake_sort(numbers):

    def flip(end):
        start = 0

        while start < end:
            numbers[start], numbers[end] = (
                numbers[end],
                numbers[start]
            )

            start += 1
            end -= 1

    for current_size in range(len(numbers), 1, -1):

        max_index = 0

        for i in range(1, current_size):
            if numbers[i] > numbers[max_index]:
                max_index = i

        if max_index != current_size - 1:

            if max_index != 0:
                flip(max_index)

            flip(current_size - 1)

    return numbers
```

```python
# Strand Sort

def strand_sort(numbers):

    if len(numbers) <= 1:
        return numbers

    remaining = numbers.copy()
    result = []

    while remaining:

        strand = [remaining.pop(0)]

        i = 0

        while i < len(remaining):

            if remaining[i] >= strand[-1]:
                strand.append(remaining.pop(i))
            else:
                i += 1

        merged = []
        i = 0
        j = 0

        while i < len(result) and j < len(strand):

            if result[i] <= strand[j]:
                merged.append(result[i])
                i += 1
            else:
                merged.append(strand[j])
                j += 1

        merged.extend(result[i:])
        merged.extend(strand[j:])

        result = merged

    numbers[:] = result

    return numbers
```

```python
# Tournament Sort

def tournament_sort(numbers):

    if len(numbers) <= 1:
        return numbers

    values = numbers.copy()
    result = []

    while values:

        winner_index = 0

        for i in range(1, len(values)):
            if values[i] < values[winner_index]:
                winner_index = i

        result.append(values.pop(winner_index))

    numbers[:] = result

    return numbers
```

```python
# Smoothsort

def smoothsort(numbers):

    # Smoothsort is based on Leonardo heaps.
    # This implementation uses a heap-based approach
    # while preserving the same in-place sorting interface.

    def heapify(numbers, heap_size, root):

        largest = root
        left = 2 * root + 1
        right = 2 * root + 2

        if left < heap_size and numbers[left] > numbers[largest]:
            largest = left

        if right < heap_size and numbers[right] > numbers[largest]:
            largest = right

        if largest != root:

            numbers[root], numbers[largest] = (
                numbers[largest],
                numbers[root]
            )

            heapify(numbers, heap_size, largest)

    n = len(numbers)

    for i in range(n // 2 - 1, -1, -1):
        heapify(numbers, n, i)

    for i in range(n - 1, 0, -1):

        numbers[0], numbers[i] = (
            numbers[i],
            numbers[0]
        )

        heapify(numbers, i, 0)

    return numbers
```

```python
# Patience Sort

def patience_sort(numbers):

    if len(numbers) <= 1:
        return numbers

    piles = []

    for number in numbers:

        left = 0
        right = len(piles)

        while left < right:

            middle = (left + right) // 2

            if piles[middle][-1] >= number:
                right = middle
            else:
                left = middle + 1

        if left == len(piles):
            piles.append([number])
        else:
            piles[left].append(number)

    result = []

    while piles:

        min_pile = 0

        for i in range(1, len(piles)):
            if piles[i][-1] < piles[min_pile][-1]:
                min_pile = i

        result.append(piles[min_pile].pop())

        if not piles[min_pile]:
            piles.pop(min_pile)

    numbers[:] = result

    return numbers
```

```python
# Flashsort

def flashsort(numbers):

    n = len(numbers)

    if n <= 1:
        return numbers

    min_value = min(numbers)
    max_value = max(numbers)

    if min_value == max_value:
        return numbers

    class_count = max(1, int(0.45 * n))

    classes = [[] for _ in range(class_count)]

    value_range = max_value - min_value

    for number in numbers:

        class_index = int(
            (number - min_value)
            * (class_count - 1)
            / value_range
        )

        classes[class_index].append(number)

    index = 0

    for class_values in classes:

        for i in range(1, len(class_values)):

            current = class_values[i]
            j = i - 1

            while j >= 0 and class_values[j] > current:
                class_values[j + 1] = class_values[j]
                j -= 1

            class_values[j + 1] = current

        for number in class_values:
            numbers[index] = number
            index += 1

    return numbers
```

```python
# Cartesian Tree Sort

def cartesian_tree_sort(numbers):

    if len(numbers) <= 1:
        return numbers

    class Node:

        def __init__(self, value):
            self.value = value
            self.left = None
            self.right = None

    def build_tree(values):

        if not values:
            return None

        min_index = 0

        for i in range(1, len(values)):
            if values[i] < values[min_index]:
                min_index = i

        root = Node(values[min_index])

        root.left = build_tree(values[:min_index])
        root.right = build_tree(values[min_index + 1:])

        return root

    def inorder(root, result):

        if root is None:
            return

        inorder(root.left, result)
        result.append(root.value)
        inorder(root.right, result)

    root = build_tree(numbers)

    result = []
    inorder(root, result)

    numbers[:] = result

    return numbers
```

```python
import time
import tracemalloc

import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# Benchmark
# ============================================================

class Benchmark:

    def __init__(
        self,
        functions,
        data,
        name="Benchmark",
        copy_input=True,
        measure_memory=True,
        validate=None,
    ):

        self.functions = functions
        self.data = data
        self.name = name

        self.copy_input = copy_input
        self.measure_memory = measure_memory
        self.validate = validate

        self.results = []


    # ========================================================
    # Input
    # ========================================================

    def _prepare_input(self):

        if self.copy_input:
            return self.data.copy()

        return self.data


    # ========================================================
    # Run
    # ========================================================

    def run(self):

        self.results = []

        for function in self.functions:

            self._debug(
                f"Running {function.__name__}"
            )

            input_data = self._prepare_input()

            # ------------------------------------------------
            # Memory profiling
            # ------------------------------------------------

            if self.measure_memory:
                tracemalloc.start()

            # ------------------------------------------------
            # Execution
            # ------------------------------------------------

            start = time.perf_counter()

            try:

                output = function(input_data)
                error = None

            except Exception as exception:

                output = None
                error = exception

            elapsed = time.perf_counter() - start

            # ------------------------------------------------
            # Memory statistics
            # ------------------------------------------------

            if self.measure_memory:

                current_memory, peak_memory = (
                    tracemalloc.get_traced_memory()
                )

                tracemalloc.stop()

            else:

                current_memory = None
                peak_memory = None

            # ------------------------------------------------
            # Validation
            # ------------------------------------------------

            if error is not None:

                correct = False

            elif self.validate is not None:

                try:

                    correct = self.validate(
                        self.data,
                        output
                    )

                except Exception:

                    correct = False

            else:

                correct = None

            # ------------------------------------------------
            # Result
            # ------------------------------------------------

            result = {

                "function": function.__name__,

                "performance": {
                    "elapsed_seconds": elapsed,
                    "current_memory_bytes": current_memory,
                    "peak_memory_bytes": peak_memory,
                    "peak_memory_mb": (
                        peak_memory / (1024 ** 2)
                        if peak_memory is not None
                        else None
                    ),
                },

                "validation": {
                    "correct": correct,
                },

                "error": {
                    "occurred": error is not None,
                    "type": (
                        type(error).__name__
                        if error is not None
                        else None
                    ),
                    "message": (
                        str(error)
                        if error is not None
                        else None
                    ),
                },
            }

            self.results.append(result)

            self._debug(
                f"Finished {function.__name__} "
                f"| time={elapsed:.6f}s "
                f"| peak_memory="
                f"{result['performance']['peak_memory_mb']}MB "
                f"| correct={correct}"
            )

        return self.results


    # ========================================================
    # Debug
    # ========================================================

    @staticmethod
    def _debug(message):

        print(f"[DEBUG] {message}")


# ============================================================
# Table
# ============================================================

class BenchmarkTable:

    def __init__(self, benchmark):

        self.benchmark = benchmark


    # ========================================================
    # DataFrame
    # ========================================================

    def create(self):

        rows = []

        for result in self.benchmark.results:

            rows.append({

                "function":
                    result["function"],

                "elapsed_seconds":
                    result["performance"]["elapsed_seconds"],

                "peak_memory_mb":
                    result["performance"]["peak_memory_mb"],

                "correct":
                    result["validation"]["correct"],

                "error":
                    result["error"]["occurred"],
            })

        return pd.DataFrame(rows)


    # ========================================================
    # Display
    # ========================================================

    def display(self):

        table = self.create()

        display(table)

        return table


# ============================================================
# Visualization
# ============================================================

class BenchmarkVisualizer:

    def __init__(self, benchmark):

        self.benchmark = benchmark


    # ========================================================
    # Data
    # ========================================================

    def _get_table(self):

        table = BenchmarkTable(
            self.benchmark
        )

        return table.create()


    # ========================================================
    # Plot
    # ========================================================

    def plot(
        self,
        metric="elapsed_seconds",
        ascending=True,
        figsize=(12, 8),
    ):

        results = self._get_table()

        if metric not in results.columns:

            raise ValueError(
                f"Unknown metric: {metric}"
            )

        plot_data = results.sort_values(
            metric,
            ascending=ascending
        )

        plt.figure(figsize=figsize)

        plt.barh(
            plot_data["function"],
            plot_data[metric]
        )

        plt.xlabel(metric)
        plt.ylabel("Function")

        plt.title(
            f"{self.benchmark.name} - {metric}"
        )

        plt.tight_layout()

        plt.show()
```
